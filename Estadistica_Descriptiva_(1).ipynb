## Carga de Datos

* ABV: La cantidad de alcohol x volumen en la cerveza

```python
datos_cervecerias =  pd.ExcelFile('/content/drive/MyDrive/Curso Estadistica con Python/Craft Beers.xlsx').parse(sheet_name='brewery', headers = 0, index_col = False)
datos_cervecerias.head()
```

```python
# Combinacion de Tablas

datos = pd.merge(datos_cervezas, datos_cervecerias,
                 left_on='brewery_id',
                 right_on = 'id',
                 how = 'inner',
                 sort = True,
                 suffixes = ('_cervezas','_cervecerias'))

datos.head()
```

## Tipo de Datos

```python
datos.info()
```

```python
def obtener_categoria(series):
  conteo_unico = series.nunique(dropna = True)
  conteo_total = series.notna().sum()

  if conteo_unico == conteo_total:
    return 'Llave Primaria'

  elif pd.api.types.is_numeric_dtype(series):
    return 'Cuantitativo'

  elif pd.api.types.is_datetime64_any_dtype(series):
    return 'Fecha'

  else:
    return 'Cualitativo'


def print_categorias(datos):
  for columna in datos.columns:
    print(f'{columna}: {obtener_categoria(datos[columna])}')

print_categorias(datos)

```

## Medidas de Tendencia Central

* Media
* Mediana
* Moda

```python
# Media (promedio)

print('El valor de la media: ', round(datos['ibu'].mean(),2))
```

```python
# Mediana

print('El valor de la mediana: ', datos['ibu'].median())
```

```python
# Moda

print('El valor de la moda: ', datos['ibu'].mode())
```

## Medidas de Dispersion

* Varianza
* Desviacion Estandar

```python
# Varianza

print('El valor de la varianza: ', round(datos['ibu'].var(),2))
```

```python
# Desviacion Estandar

print('El valor de la desviacion estandar: ', round(datos['ibu'].std(),2))
```

## Medidas de Rango

* Minimo
* Maximo
* Rango

```python
# Minimo

print('El valor del minimo: ', datos['ibu'].min())
```

```python
# Maximo

print('El valor del maximo: ', datos['ibu'].max())
```

```python
# Rango

print('El valor del rango: ', datos['ibu'].max() - datos['ibu'].min())
```

## Medidas de Cuantiles


* Percentil
* Decil
* Cuartil
* Quintil

```python
# Percentil

print('El valor del percentil 83: ', datos['ibu'].quantile(0.83))
```

```python
# Decil

print('El valor del decil 4: ', datos['ibu'].quantile(0.4))
```

```python
# Cuartiles

print('El valor del primer cuartil: ', datos['ibu'].quantile(0.25))
```

```python
# Quintil

print('El valor del cuarto quintil: ', datos['ibu'].quantile(0.8))
```

## Covarianza

```python
covarianza = datos['ibu'].cov(datos['abv'])

print('La covarianza es: ', covarianza)
```

```python
cov_matrix = datos.cov(numeric_only= True)

print('La covarianza es: \n', cov_matrix)
```

## Correlacion

```python
# Pearson

datos[['ibu','abv','ounces']].corr(method = 'pearson')
```

```python
# Spearman Rho

datos[['ibu','abv','ounces']].corr(method = 'spearman')
```

```python
# Tau Kendall

datos[['ibu','abv','ounces']].corr(method = 'kendall')
```
