## Proyecto Final: Análisis de Ventas Retail (Superstore)
Este proyecto analiza el desempeño de ventas de una cadena de retail, identificando patrones temporales, productos líderes y comportamiento regional.

**Etapa 1: Búsqueda (Completada)**

Archivos seleccionados: orders_data_01.csv (Transacciones) y returns_data_02.csv (Devoluciones).

Justificación: Permiten realizar uniones (cruces) por Order ID y contienen variables temporales y categóricas para el análisis.

**Etapa 2: Carga y Exploración**

En esta etapa importamos las librerías esenciales y cargamos la data cruda. Usamos pandas para manipulación y visualización.

```python
# a) Importación de paquetes necesarios
import pandas as pd
import numpy as np
import plotly.express as px  # Para gráficos interactivos
import plotly.graph_objects as go
```

```python
# b) Carga de bases de datos
# Comentario: Usamos sep=';' porque los archivos CSV originales usan punto y coma como delimitador.
df_orders = pd.read_csv('orders_data_01.csv', sep=';', encoding='latin1')
df_returns = pd.read_csv('returns_data_02.csv', sep=';', encoding='latin1')
```

```python
# c) 5 Técnicas de exploración
print("--- 1. Forma del DataFrame (Filas, Columnas) ---")
print(f"Orders: {df_orders.shape}")
print(f"Returns: {df_returns.shape}")
```

```python
print("\n--- 2. Tipos de datos ---")
print(df_orders.dtypes)
```

```python
print("\n--- 3. Primeras filas (Validación de carga) ---")
print(df_orders.head(3))
```

```python
print("\n--- 4. Estadísticos descriptivos básicos ---")
print(df_orders.describe())
```

```python
print("\n--- 5. Verificación de Nulos iniciales ---")
print(df_orders.isnull().sum())
```

**Etapa 3: Manipulación**

Aquí transformamos los datos para hacerlos útiles. Convertimos textos a fechas reales y unimos las tablas.

```python
# a) Selección de variables
# Comentario: Creamos un subconjunto con solo las columnas necesarias para reducir uso de memoria.
cols_utiles = ['Order ID', 'Order Date', 'Customer Name', 'Segment', 'Region',
               'Category', 'Sub-Category', 'Sales', 'Quantity', 'Profit']
df_clean = df_orders[cols_utiles].copy()
```

```python
# b) Limpiar caracteres especiales
# Comentario: Aunque 'Sales' ya vino limpio, este código asegura eliminar '$' o ',' si existieran
# y convierte la columna a float, cumpliendo el requisito de limpieza de moneda.
if df_clean['Sales'].dtype == 'object':
    df_clean['Sales'] = df_clean['Sales'].astype(str).str.replace('$', '').str.replace(',', '').astype(float)
```

```python
# c) Formato de fecha y conversión
# Comentario: pd.to_datetime convierte el texto a objeto fecha real. 'dayfirst=True' es vital para fechas tipo 31/12/2020.
df_clean['Order Date'] = pd.to_datetime(df_clean['Order Date'], dayfirst=True)
```

```python
# Requisito específico: Crear columna con formato string '%Y-%m-%d %H:%M:%S'
df_clean['Fecha_Formato_ISO'] = df_clean['Order Date'].dt.strftime('%Y-%m-%d %H:%M:%S')
```

```python
# d) Extracción de fecha (Ingenieria de Caracteristicas)
# Comentario: Extraemos componentes de la fecha para poder agrupar por mes o año más adelante.
df_clean['Año'] = df_clean['Order Date'].dt.year
df_clean['Mes'] = df_clean['Order Date'].dt.month
df_clean['Mes_Nombre'] = df_clean['Order Date'].dt.month_name()
```

```python
# e) Combinación de bases de datos (Merge)
# Comentario: Usamos 'left join'. Mantenemos todas las ventas de 'df_clean' y traemos información
# de 'df_returns' solo si hay coincidencia en 'Order ID'.
df_final = pd.merge(df_clean, df_returns, on='Order ID', how='left')
```

```python
print("Dimensiones después del merge:", df_final.shape)
print(df_final.head(21))
```

**Etapa 4: Limpieza**

Tratamos los valores vacíos generados por el cruce y buscamos duplicados.

```python
# a) Consulta de valores vacíos
print("\n--- Valores Vacíos por Columna ---")
print(df_final.isnull().sum())
```

```python
# b) Reemplazar vacíos
# Comentario: Al hacer el merge, las ventas NO devueltas tienen NaN en la columna 'Returned'.Rellenamos estos NaNs con "No", ya que significa que no hubo devolución.
df_final['Returned'] = df_final['Returned'].fillna('No')

# Si hubiera vacíos en columnas numéricas (ej. Quantity), usaríamos la media:
# df_final['Quantity'] = df_final['Quantity'].fillna(df_final['Quantity'].mean())
```

```python
# c) Validar duplicados
# Comentario: Buscamos filas idénticas. Si existen, las eliminamos para no inflar las ventas.
duplicados = df_final.duplicated().sum()
print(f"\nSe encontraron {duplicados} filas duplicadas.")

if duplicados > 0:
    df_final = df_final.drop_duplicates()
    print("Duplicados eliminados.")
```

**Etapa 5: Simplificar (Agrupaciones)**

Creamos tablas resumen (pivots) para responder preguntas de negocio.

```python
# Agrupación 1: Tendencia Temporal (Año y Mes)
# Comentario: Sumamos Ventas y Cantidad agrupando por Año y Mes. Reseteamos índice para que sea DataFrame plano.
agrupacion1 = df_final.groupby(['Año', 'Mes'])[['Sales', 'Quantity']].sum().reset_index()
agrupacion1 = agrupacion1.sort_values(['Año', 'Mes'])
```

```python
agrupacion1.head()
```

```python
# Agrupación 2: Desempeño por Producto
# Comentario: Agrupamos por Categoría y Subcategoría para ver qué productos venden más.
agrupacion2 = df_final.groupby(['Category', 'Sub-Category'])[['Sales', 'Quantity']].sum().reset_index()
agrupacion2 = agrupacion2.sort_values('Sales', ascending=False)
```

```python
agrupacion2.head()
```

```python
# Agrupación 3: Desempeño Regional
# Comentario: Analizamos las ventas por Región y Segmento de cliente.
agrupacion3 = df_final.groupby(['Region', 'Segment'])['Sales'].sum().reset_index()
agrupacion3 = agrupacion3.sort_values('Sales', ascending=False)
```

```python
agrupacion3.head()
```

```python
print("\n--- Top 5 Meses con más ventas ---")
print(agrupacion1.sort_values('Sales', ascending=False).head())
```

**Etapa 6: Visualización (Con Plotly)**

A continuación, haremos gráficos dinamicos usando la libreria que se menciona lineas arriba.

**Gráfico 1: Evolución de Ventas Mensuales (Line Chart)**

Mostraremos la estacionalidad de las ventas. Podremos ver picos al final de cada año.

```python
fig1 = px.line(agrupacion1,
               x='Mes',
               y='Sales',
               color='Año',
               markers=True,
               title='Evolución de Ventas Mensuales por Año',
               labels={'Sales': 'Ventas ($)', 'Mes': 'Mes del Año'})
fig1.show()
```

**Conclusión:**

Se observa una fuerte estacionalidad. Las ventas tienden a dispararse en los últimos meses del año (Noviembre-Diciembre), probablemente debido a festividades sin dejar de mencionar algunas fluctuaciones a la baja algo marcada en el año 2017.

**Gráfico 2: Ventas por Categoría y Subcategoría (Bar Chart)**

Permite identificar los productos "estrella" y los que menos aportan.

```python
fig2 = px.bar(agrupacion2,
              x='Sales',
              y='Sub-Category',
              color='Category',
              orientation='h',
              title='Ranking de Ventas por Subcategoría y Categoría',
              text_auto='.2s', # Muestra el valor en la barra
              labels={'Sales': 'Ventas Totales ($)', 'Sub-Category': 'Producto'})
fig2.update_layout(yaxis={'categoryorder':'total ascending'}) # Ordena de mayor a menor
fig2.show()
```

**Conclusión:**

 Los productos tecnológicos (Phones) y Sillas (Chairs) son los mayores generadores de ingresos. Categorías como "Fasteners" o "Labels" tienen un impacto marginal en la facturación total.

**Gráfico 3: Ventas por Región y Segmento (Sunburst / Bar Chart)**

Analizamos graficamente dónde están nuestros mejores clientes.

```python
fig3 = px.bar(agrupacion3,
              x='Region',
              y='Sales',
              color='Segment',
              barmode='group',
              title='Comparativa de Ventas por Región y Segmento',
              labels={'Sales': 'Ventas ($)'})
fig3.show()
```

**Conclusión:**

La región "West" lidera las ventas, impulsada fuertemente por el segmento "Consumer" (Consumidor final). El segmento "Home Office" es el que menos volumen genera en todas las regiones.

##**Capítulo 1:**

El Hallazgo (Etapas 1 y 2) Nos sumergimos en los archivos obtenidos de la compañía y rescatamos dos piezas clave del rompecabezas:

El Diario de Ventas (Orders): Una bitácora con casi 10,000 transacciones entre 2014 y 2017.

El Reporte de Fallos (Returns): Una lista aislada de pedidos que habían sido devueltos por los clientes.

Al cargar estos datos a nuestro entorno de análisis (Python-colab), descubrimos que, aunque ricos en información, eran "mudos" por separado. Las fechas eran simples textos y no había conexión entre lo que se vendía y lo que se devolvía.

##**Capítulo 2:**

La Transformación (Etapa 3) Aquí actuamos como traductores de la data.

Primero, estandarizamos el tiempo: Convertimos las fechas de texto plano a objetos de tiempo real (datetime), permitiéndonos por primera vez "ver" los años y meses.

Segundo, construimos el puente: Realizamos un cruce de datos (Left Merge). Tomamos cada una de las 10,000 ventas y le "preguntamos" a la lista de devoluciones: "¿Este pedido regresó?". Esto enriqueció nuestra tabla principal, agregando una dimensión crítica de éxito o fracaso a cada transacción.

##**Capítulo 3:**

La Limpieza (Etapa 4) Al unir las dos datas en formato csv, aparecieron los vacíos. Las ventas exitosas (que no fueron devueltas) generaron espacios vacíos (Nulls) en la columna de devoluciones.

Lejos de ver esto como un error, aplicamos lógica de negocio: Rellenamos esos vacíos con un rotundo "No", confirmando que esas ventas fueron efectivas.

También aseguramos la integridad financiera eliminando duplicados y limpiando cualquier símbolo extraño en los montos de dinero, garantizando que cada dólar sumado fuera real

##**Capítulo 4:**

La Revelación (Etapas 5 y 6) Con la data limpia y unificada, simplificamos la información en tres grandes verdades visuales:

El Ritmo del Negocio: Descubrimos una estacionalidad agresiva. La empresa "duerme" a principios de año y despierta violentamente en los últimos meses (noviembre-diciembre).

Los Reyes del Catálogo: No todos los productos son iguales. Las Sillas (Chairs) y los Teléfonos (Phones) son los verdaderos protagonistas de los ingresos. Vender sujetadores o etiquetas es casi irrelevante para la facturación final.

El Bastión del Oeste: Geográficamente, la región West (Oeste) es nuestro motor principal, impulsada masivamente por el Consumidor Final, quien supera por mucho a las compras corporativas.
